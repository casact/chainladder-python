# Sahasrabuddhe verification

```python
import chainladder as cl

genins = cl.load_sample("genins")
genins
```

```python
# Limits
triangle_limit = 1000000
basic_limit = 500000
policy_limit = 2000000
```

```python
# B1: Exposure Period Trend Index [ 2% EP Trend; 5% between EP 6 and 7 ]
#
# "a one-time increase TO 5%" replaces the 2% for the single EP 6 -> EP 7 step
# rather than stacking on top of it, so the assumption is three segments of
# annual trend rather than one rate plus a bump.
#
# Two arguments do the work that used to need a carrier triangle of 1s:
#
#   full_triangle=True  returns the factor surface over the whole 10 x 10
#                       origin x development rectangle.  The exhibits are full
#                       rectangles, but genins is a triangle, so without this
#                       everything below the latest diagonal comes back NaN.
#
#   base_period=2001    anchors the factors on EP 1 instead of the default
#                       (the latest origin).  The paper's index runs EP 1 =
#                       1.000 and rises; the default would put 1.000 at EP 10.
#
# `trend_` is a multiplier TO the base period's cost level -- it answers "what
# do I multiply EP i by to state it at EP 1's level".  A cost level INDEX runs
# the other way, so the index is its reciprocal.
B1 = (
    1
    / cl
    .Trend(
        trends=[0.02, 0.05, 0.02],
        dates=[
            ("2006-12-31", None),  # 2% from the earliest origin up to EP 6
            ("2007-12-31", "2006-12-31"),  # 5% for the single EP 6 -> EP 7 step
            ("2010-12-31", "2007-12-31"),  # 2% from EP 7 through the latest origin
        ],
        axis="origin",
        base_period=2001,
        full_triangle=True,
    )
    .fit(genins)
    .trend_
)

B1
```

```python
cl.Trend(
    trends=[0.02, 0.05, 0.02],
    dates=[
        ("2006-12-31", None),  # 2% from the earliest origin up to EP 6
        ("2007-12-31", "2006-12-31"),  # 5% for the single EP 6 -> EP 7 step
        ("2010-12-31", "2007-12-31"),  # 2% from EP 7 through the latest origin
    ],
    axis="origin",
    base_period=2001,
    full_triangle=True,
).fit(genins).trend_
```

```python
# B2: Calendar Period Trend Index [ 1% Calendar Period Trend; -5% between CP 2 and 3 ]
#
# Same shape of assumption as B1, but on the valuation axis.  The index
# therefore depends only on calendar period k = i + j - 1, which is why the
# result is constant down each diagonal rather than across each row.
#
# CP 1 is 2001, so the one-time -5% sits on the 2002 -> 2003 valuation step,
# and base_period is the 2001 valuation rather than the 2001 origin.  On the
# valuation axis a base period names a month, so "2001-12" rather than 2001 --
# a bare year would resolve to January 2001, which is not a valuation date here.
#
# The last segment leaves `start` as None.  Under full_triangle=True that
# defaults to the rectangle's own last valuation (2020) rather than genins'
# valuation date (2010), which is what lets the factors keep accruing across
# the whole exhibit instead of flattening at the latest diagonal.
B2 = (
    1
    / cl
    .Trend(
        trends=[0.01, -0.05, 0.01],
        dates=[
            ("2002-12-31", None),  # 1% from the earliest valuation up to CP 2
            ("2003-12-31", "2002-12-31"),  # -5% for the single CP 2 -> CP 3 step
            (None, "2003-12-31"),  # 1% from CP 3 out to the last valuation
        ],
        axis="valuation",
        base_period="2001-12",
        full_triangle=True,
    )
    .fit(genins)
    .trend_
)

B2
```

```python
# B3: Combined Trend Index [ B1 * B2 ]
#
# Exposure period and calendar period trend compound, so the combined cost
# level index is the elementwise product.  This is the T(i, j) that equation
# 3.1 uses to shift the claim size model parameters, and it is the first index
# that varies in BOTH directions -- B1 is constant across j, B2 is constant
# down each diagonal, and the product is constant in neither.
#
# Both operands are full rectangles anchored on the same cell, so the product
# needs no further rebasing: cell (1, 1) is simultaneously EP 1 and CP 1.
B3 = B1 * B2

B3
```

```python
# Check the reconstruction against the exhibits as printed in the paper.
#
# The published grids carry three decimals, so agreement is asserted to half of
# the last printed digit.  Two consequences of that rounding are worth naming:
#
#   * B2 cell (1, 3) is exactly 1.01 * 0.95 = 0.9595.  The paper rounds half up
#     and prints 0.960; numpy rounds the float (which sits a hair below .9595)
#     down and shows 0.959.  Same number, different tie-break.
#
#   * B3 as printed is NOT the product of B1 and B2 as printed -- the paper
#     multiplied unrounded values, so the two disagree by up to 0.0013.  B3 is
#     therefore checked against its own published grid, not against B1 * B2.
import numpy as np

B1_published = np.tile(
    np.array([1.000, 1.020, 1.040, 1.061, 1.082, 1.104, 1.159, 1.182, 1.206, 1.230])[
        :, None
    ],
    (1, 10),
)

# B2 is constant along diagonals, so 19 calendar-period values determine it:
# exhibit row 1 gives k = 1..10, exhibit column 10 gives k = 10..19.
_cp = np.array([
    1.000,
    1.010,
    0.960,
    0.969,
    0.979,
    0.989,
    0.998,
    1.008,
    1.019,
    1.029,
    1.039,
    1.049,
    1.060,
    1.070,
    1.081,
    1.092,
    1.103,
    1.114,
    1.125,
])
_i, _j = np.indices((10, 10))
B2_published = _cp[_i + _j]

B3_published = np.array([
    [1.000, 1.010, 0.960, 0.969, 0.979, 0.989, 0.998, 1.008, 1.019, 1.029],
    [1.030, 0.979, 0.988, 0.998, 1.008, 1.018, 1.029, 1.039, 1.049, 1.060],
    [0.998, 1.008, 1.018, 1.029, 1.039, 1.049, 1.060, 1.070, 1.081, 1.092],
    [1.028, 1.039, 1.049, 1.060, 1.070, 1.081, 1.092, 1.103, 1.114, 1.125],
    [1.059, 1.070, 1.081, 1.092, 1.102, 1.114, 1.125, 1.136, 1.147, 1.159],
    [1.091, 1.102, 1.113, 1.125, 1.136, 1.147, 1.159, 1.170, 1.182, 1.194],
    [1.157, 1.169, 1.181, 1.193, 1.204, 1.217, 1.229, 1.241, 1.253, 1.266],
    [1.192, 1.204, 1.216, 1.229, 1.241, 1.253, 1.266, 1.278, 1.291, 1.304],
    [1.228, 1.241, 1.253, 1.266, 1.278, 1.291, 1.304, 1.317, 1.330, 1.344],
    [1.266, 1.278, 1.291, 1.304, 1.317, 1.330, 1.343, 1.357, 1.370, 1.384],
])

for name, computed, published in [
    ("B1", B1, B1_published),
    ("B2", B2, B2_published),
    ("B3", B3, B3_published),
]:
    deviation = np.abs(computed.values[0, 0] - published).max()
    assert deviation <= 0.0006, f"{name} deviates from the paper by {deviation:.6f}"
    print(f"{name}: matches the published exhibit (max deviation {deviation:.6f})")
```

```python
# Applying a trend to the values is just multiplication -- multiplying by a
# factor triangle and calling .trend() on the data give identical answers.
# Materializing the factors is only worth it when you want to inspect or reuse
# them, as here.
#
# B3 is a cost level INDEX (cell (1, 1) = 1.000), so it is not itself the thing
# you multiply by.  To restate every cell to exposure period n's cost level --
# the T(n, j) / T(i, j) ratio that equation 3.2 needs -- divide exposure period
# n's row by the whole index.  Taking a row rather than a single cell matters
# here and not for B1: B3 varies by development interval as well as by origin.
to_current = B3.iloc[0, 0, -1:, :] / B3

# The factors cover the full rectangle, but genins does not, so the product
# masks back to genins' own valuation date.  That is correct -- there is no
# data out there to restate.
genins * to_current
```

```python
# C1: Claims Size Model Parameters at Exposure Year 10 Cost Level
#     [ via claim size modeling ]
#
# These are an INPUT, not something derived -- section 2.2 assumes the actuary
# has already fitted a claim size model, and section 1.2 puts that fitting
# outside the paper's scope.  Example 1 uses an exponential purely for ease of
# presentation, which is the model cl.LEV implements.
#
# cl.LEV takes the result of that analysis as `means`, either as a mapping of
# development age to mean claim size or as a Triangle.  Given no `trend`, the
# mapping is laid onto the base exposure period and handed back unrestated --
# which is C1, anchored on the 2010 row because C1 is stated at exposure year
# 10's cost level.
theta_10 = {
    12: 28138,
    24: 84242,
    36: 133998,
    48: 182460,
    60: 204649,
    72: 228245,
    84: 252830,
    96: 265063,
    108: 275707,
    120: 280000,
}

C1 = cl.LEV(means=theta_10).fit(genins).means_
C1
```

```python
# C2: Claims Size Model Parameters [ C1 * B3(i,j) / B3(10,j) ]
#
# Equation 3.1.  C1 holds one row of parameters, all at exposure year 10's cost
# level.  Every other (i, j) cell sits at a different cost level, and B3 is
# exactly the index that says by how much -- so restating C1 onto the whole
# grid is C1 scaled by the ratio of each cell's index to row 10's.
#
# This is where this notebook differs from layers.ipynb.  There the restatement
# was written out as arithmetic on C1:
#
#     C2 = C1 * B3 / B3.iloc[0, 0, -1:, :]
#
# cl.LEV now owns it.  Passing `trend` makes `fit` do the scaling internally,
# so C2 comes back from a single call, and the row to divide by is named by
# `base_period` rather than by an .iloc into the index.  It defaults to the
# latest origin -- the same row the mapping was laid on -- so C1 and C2 agree
# on their base by construction instead of by the caller keeping two positional
# slices in step.
#
# This works because the exponential is a scale family: inflating claim sizes
# by a factor inflates the mean by the same factor, so adjusting the parameter
# for cost level is plain multiplication.  Section 2.2 requires exactly this
# ("that claims size model parameters can be adjusted for the impact of
# inflation"), and it is what limits the method to models of that shape.
#
# Row 10 reproduces C1 by construction: B3(10,j) / B3(10,j) is 1.
lev = cl.LEV(means=theta_10, trend=B3).fit(genins)

C2 = lev.means_
C2
```

```python
# C3: Unlimited Means
#
# For an exponential model the mean IS the parameter, so C3 restates C2 without
# changing it.  The paper prints it as its own exhibit because the general
# method needs M(Phi) -- the unlimited first moment -- as a distinct quantity
# from the parameters: for a lognormal or Pareto the two would differ, in
# dimension (Phi is a vector) and in existence (a Pareto with alpha <= 1 has
# no finite mean while its parameters are perfectly well defined).
#
# The limit is a property of the quantity being asked for, so it is named at
# construction: each exhibit in the C series is its own cl.LEV, and `lev_`
# carries the answer.  np.inf is the default, so C3 is also `lev.lev_` off the
# C2 instance above -- spelled out in full here to keep the series uniform.
C3 = cl.LEV(means=theta_10, trend=B3, limit=np.inf).fit(genins).lev_
C3
```

```python
# C4: Limited Expected Values at Policy Limits
#
# LEV(d) = theta * (1 - exp(-d / theta)), capping each claim at the policy
# limit.  At 2,000,000 the cap bites only slightly -- compare against C3 and
# the early development ages are unchanged to the printed precision, because a
# mean of ~22,000 puts almost no probability above two million.
C4 = cl.LEV(means=theta_10, trend=B3, limit=policy_limit).fit(genins).lev_
C4
```

```python
cl.LEV(means=theta_10, trend=B3, limit=policy_limit).fit(genins).means_
```

```python
# C5: Limited Expected Values at Limits of Data Triangle
#
# The same calculation at the limit the observed data is actually capped at.
# This is the denominator side of the limit adjustment: genins is reported at
# the triangle limit, so its development pattern is a triangle-limit pattern,
# and C5 is what that implies about expected claim size.
C5 = cl.LEV(means=theta_10, trend=B3, limit=triangle_limit).fit(genins).lev_
C5
```

```python
# C6: Limited Expected Values at Basic Limit
#
# Lowest of the three caps, so the largest gap from the unlimited mean -- by
# development age 120 the basic limit retains 233,050 of a 280,000 mean.
#
# With C4, C5 and C6 in hand the limit adjustment factors of equations 2.1/2.2
# follow directly, since each is a ratio of layer expectations.  A layer is
# itself just another cl.LEV: `attachment` and `limit` together give
# LEV(p) - LEV(d), which is equation 2.4.  E2 and E3 below are those layers.
C6 = cl.LEV(means=theta_10, trend=B3, limit=basic_limit).fit(genins).lev_
C6
```

```python
# cl.LEV is a transformer now, which is what lets the claim size model travel
# with the data rather than beside it.
#
# `transform` follows the library convention -- it copies the triangle it is
# given and attaches the fitted quantities, leaving the values alone:
lev_transformed = lev.transform(genins)

assert (lev_transformed == genins).all(), "transform leaves the values untouched"
assert np.allclose(
    np.nan_to_num(lev_transformed.means_.values),
    np.nan_to_num(C2.values),
), "the restated parameters ride along on the triangle"
assert np.allclose(
    np.nan_to_num(lev_transformed.lev_.values),
    np.nan_to_num(C3.values),
), "so does the layer expectation, here the unlimited mean"

print("transform attaches the fitted quantities and leaves the data alone")
print(f"  values unchanged : {(lev_transformed == genins).all()}")
print(f"  carries means_   : {hasattr(lev_transformed, 'means_')}")
print(f"  carries lev_     : {hasattr(lev_transformed, 'lev_')}")

# Which also means it composes.  Nothing downstream here consumes those off a
# triangle -- cl.Sahasrabuddhe builds its own cl.LEV from the same arguments --
# but a step that did would find them where the library puts every other
# fitted quantity.
cl.Pipeline([("lev", cl.LEV(means=theta_10, trend=B3))]).fit_transform(genins)
```

```python
# D: Calculation of Development Factors at Basic Limit
#
# Sections A to C worked the restatement by hand, exhibit by exhibit, to show
# where each number comes from and to check it against the paper.  This is the
# same calculation as it would actually be written.
#
# Everything above collapses into one call.  cl.Sahasrabuddhe takes the claim
# size analysis and the cost level index and restates the triangle onto a
# single basis; `triangle_` is that restated data, and ordinary cl.Development
# fits on it.  There is nothing left in the triangle for the factors to be
# contaminated by, so no bespoke development code is needed -- and a tail,
# weights, or exclusions would follow as ordinary steps without any of this
# changing.
#
# Three limits are named, and they play different parts:
#
#   data_limit    where the observed data is capped -- the denominator of
#                 equation 3.2, varying cell by cell with cost level.
#
#   basic_limit   the common basis, B.  The triangle is restated onto it.
#
#   target_layer  the layer wanted out, as (attachment, exhaustion).
#                 Restating a triangle IS the move onto the basis, so here it
#                 has to be (0, basic_limit); naming any other layer raises
#                 rather than quietly doing something else.
#                 Section E below is where it starts to differ, because there
#                 what gets fitted is a pattern rather than a triangle.
#
# No C1 through C6 are passed.  The limits are named instead and the limited
# expected values derived from them, so there is no way to pair a C5 and a C6
# that were built at different limits.
#
# No base_period either.  It defaults to the latest origin, which is where a
# mapping of means is laid, so the two agree by construction.
basis = cl.Sahasrabuddhe(
    means=theta_10,
    trend=B3,
    data_limit=triangle_limit,
    basic_limit=basic_limit,
    target_layer=(0, basic_limit),
).fit(genins)

# D1: the triangle restated to exposure year 10 cost levels and basic limit.
D1 = basis.triangle_

development = cl.Development(average="volume").fit(D1)

D1
```

```python
# D2: Exposure Year 10 Incremental Basic Limit Development Factors
#     [ per D1; Volume Weighted Averages ]
#
# Ordinary age-to-age factors off the restated triangle.  "Volume weighted
# average" is chainladder's `average="volume"`, which is also its default.
#
# These factors are the deliverable: basic limit development at exposure year
# 10's cost level, estimated from data that was neither.
D2 = development.ldf_
D2
```

```python
# D3: Exposure Year 10 Cumulative Development Factors [ per D2 ]
#
# The age-to-age factors chained from the right.
#
# chainladder stops at 108-Ult because no tail has been fitted, so the final
# age develops by the identity; the paper prints that as a tenth column of
# 1.000.  In production a tail step would follow `dev` in the Pipeline above,
# which is also what would populate it -- it is left off here so the factors
# stay comparable with the published exhibit.
D3 = development.cdf_
D3
```

```python
# E: Calculation of Development Factors by Layer
#
# D3 is one pattern: basic limit, exposure year 10 cost level.  Section 3 says
# a pattern for any OTHER layer follows from it by a ratio of layer
# expectations -- no refitting, no second triangle.  For a layer whose expected
# loss is L(i,j):
#
#     factor(i,j) = D3(j) * [ L(i,10) / C6(10,10) ] / [ L(i,j) / C6(10,j) ]
#
# L(i,10) / L(i,j) is how much of exposure period i's ultimate loss in that
# layer has emerged by age j, and the C6(10,.) terms put numerator and
# denominator on D3's own basis so the adjustment is dimensionless.
#
# This is the second thing cl.Sahasrabuddhe does, and which one it does is
# decided by what it is fitted on.  Section D fitted a triangle and got
# `triangle_`; fitting the development step instead gets the layer:
#
#   full_cdf_  the factors over the whole origin x development rectangle
#   full_ldf_  the same, age to age -- one row per cost level
#   cdf_       its latest diagonal, as an ordinary development pattern
#   ldf_       the same, age to age
#
# The target layer is named at construction as a pair of limits.  A
# ground-up layer attaches at zero: (0, basic_limit).
#
# The pattern is aligned to the rectangle's development ages by age, not by
# position.  That matters the moment a tail is fitted: cdf_ then has more
# entries than the triangle has ages, and zipping them positionally would
# quietly misalign every factor.

# E1: Basic Limit [ D3j * (C6i,10/C610,10) / (C6i,j/C610,j) ]
#
# Targeting the limit the pattern was fitted on, so this is the basic limit.
# Row 10 reproduces D3 exactly -- every ratio collapses to 1 -- and the other
# rows differ from it only by cost level.
# `data_limit` is not passed: it says where a triangle of claims is capped, and
# no claims are being restated here.  Only the target layer varies between the
# exhibits below, so that is all the helper takes.
def layer(target_layer):
    return cl.Sahasrabuddhe(
        means=theta_10,
        trend=B3,
        basic_limit=basic_limit,
        target_layer=target_layer,
    ).fit(development)


E1 = layer(target_layer=(0, basic_limit)).full_cdf_
E1
```

```python
# E1 as a development pattern.
#
# `full_cdf_` above is the full rectangle: nothing in the formula needs the
# data -- it is C6 and D3 all the way down, both defined for every (i, j) --
# so there is no reason to mask it to the triangle.  Each ROW is one exposure
# period's complete basic limit pattern, age 1 through 10, at that period's own
# cost level, and the cells below the latest diagonal are the factors a future
# valuation would use.
#
# Two ways to read the rectangle:
#
#   * Down a column: how the age-j-to-ultimate factor varies with cost level.
#     Column 1 runs 12.63, 12.54, 13.23, ..., 12.29 -- not monotonic, because
#     B3 is not monotonic in i.  The -5% calendar period step in B2 pushes the
#     early exposure periods around.
#
#   * Along the last column: every entry is 1.000 by construction, since
#     E1(i,10) = D3(10) * r(i,10)/r(i,10) regardless of cost level.
#
# `cdf_` collapses that to the ONE cell per row that each exposure period has
# actually reached -- the latest diagonal -- which is the pattern you would
# apply.  Reading a single row of the rectangle instead would apply one period's
# cost level to every other period.
layer(target_layer=(0, basic_limit)).cdf_
```

```python
# E2: Basic Limit to Policy Limit
#     [ D3j * ((C4i,10-C6i,10)/C610,10) / ((C4i,j-C6i,j)/C610,j) ]
#
# C4 - C6 is the expected loss in the layer from the basic limit up to the
# policy limit.  Naming those two limits as the target attachment and
# exhaustion is what asks for it.
#
# The first column is enormous, and genuinely so: at age 12 almost nothing has
# emerged in a layer attaching at 500,000, so the factor needed to develop it
# to ultimate is astronomical.  The paper prints "very large" rather than a
# number.  This is the substantive point of developing by layer -- excess
# layers develop far more slowly than the ground-up data suggests.
E2 = layer(target_layer=(basic_limit, policy_limit)).full_cdf_
E2
```

```python
# E3: Policy Limit to Unlimited
#     [ D3j * ((C3i,10-C4i,10)/C610,10) / ((C3i,j-C4i,j)/C610,j) ]
#
# Everything above the policy limit -- the unlimited mean less what the policy
# limit retains -- so the layer exhausts at infinity.
#
# The first two columns come back non-finite here rather than merely large.
# That is not a failure of the arithmetic: at the earliest ages the expected
# loss above 2,000,000 underflows to zero in double precision, and the paper
# writes "very large" in the same cells for the same reason.
E3 = layer(target_layer=(policy_limit, np.inf)).full_cdf_
E3
```

```python
# E4: Limit of Data in Triangle [ D3j * (C5i,10/C610,10) / (C5i,j/C610,j) ]
#
# The layer the observed data actually sits in: ground up to the triangle
# limit.  This is the one to compare against, because it is the pattern the
# method says genins should have -- stated for the limit genins is reported at,
# rather than for the basic limit.
#
# E7 puts it next to what an unadjusted chainladder would have produced.
E4 = layer(target_layer=(0, triangle_limit)).full_cdf_
E4
```

```python
# E5: Unadjusted Incremental Development Factors at Limits of Data Triangle
#     [ per A1; Volume Weighted Averages ]
#
# The naive answer: fit development straight off the raw triangle, with no
# trend adjustment and no limit adjustment at all.  Everything from B1 onward
# exists to establish whether this is good enough.
unadjusted = cl.Development(average="volume").fit(genins)

E5 = unadjusted.ldf_
E5
```
