# Triangle Visualization

Chainladder's {class}`~chainladder.Styler` class enables visual customization of triangles when displayed in a Jupyter notebook. Any styling options that can be applied to a Pandas DataFrame can also be applied to a Chainladder `Triangle`. In addition, the {class}`~chainladder.Styler` extends its [Pandas counterpart](https://pandas.pydata.org/docs/user_guide/style.html) with methods designed for actuaries.

```{note}
The {class}`~chainladder.Styler` only works on single triangles.
```

## Highlighting the lower triangle

The lower triangle contains cells that extend beyond the latest diagonal. This area can be highlited with `Triangle.style.highlight_lower_triangle()`.

````python
import chainladder as cl

raa = cl.load_sample("raa")

raa.style.highlight_lower_triangle(color="blue")
````

You may find that the blue contrasts too sharply against the white background of this website. You can change the color to a lighter shade of blue, and the text to white to improve readability.

````python
full = cl.Chainladder().fit(raa).full_triangle_
full.style.highlight_lower_triangle(
    color="#2F61AD",
    text_color="#FFFFFF",
)
````

## Highlighting diagonals

Diagonals represent cells sharing the same valuation date. The latest diagonal can be highlighted with `Triangle.style.highlight_diagonal()`. You can also specify a particular valuation date, or chain diagonal and lower triangle styling together.

````python
# Highlight the latest diagonal and the lower triangle together
raa.style.highlight_diagonal(color="#FFE599").highlight_lower_triangle(
    color="#DDEBF7"
)
````

## Numeric Formatting

The `Styler` allows you to control the numeric format of a `Triangle`'s values. Let's change the precision of the link ratios to have 3 decimal places.

````python
raa.link_ratio.style.format(
    precision=3,
    na_rep="",
)
````

## Localization

Chainladder is used by actuaries all over the world. The `Styler` can be used to adapt the numeric formatting to specific locales. Let's try formating `genins` with the Swiss numbering format, which uses apostrophes as the thousands separator.

````python
genins = cl.load_sample("genins")

genins.style.format(
    precision=0,
    thousands="'",
    na_rep="",
)
````

### Babel

You can supply a format using the [babel](https://babel.pocoo.org/en/latest/) package if you have it installed. Let's try the [Indian numeric format](https://en.wikipedia.org/wiki/Indian_numbering_system).

````python
from babel.numbers import format_decimal

# The Indian numbering system groups by lakh and crore: 12,34,567.
genins.style.format(
    lambda x: format_decimal(x, locale="en_IN"),
    na_rep="",
)
````
